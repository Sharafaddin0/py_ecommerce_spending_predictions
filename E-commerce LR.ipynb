```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Ecommerce Customers.csv')
df.head()
```

```python
# Get some statistical information about the dataset
df.describe()
```

```python
df.info()
```

```python
# Exploratory Data Analysis of annual time spending on WEBSITE
sns.jointplot(x = 'Time on Website' , y = 'Yearly Amount Spent', data = df, alpha = 0.5)
```

```python
# Exploratory Data Analysis of annual time spend on APPLICATION
sns.jointplot(x = 'Time on App', y = 'Yearly Amount Spent', data = df, color = 'red', alpha = 0.3)
```

```python
# Pairplot of all numeric values
sns.pairplot(df, kind = 'scatter', plot_kws = {'color': 'blue', 'alpha' : 0.4})
```

```python
# Split the data into training and test sets
from sklearn.model_selection import train_test_split
X = df[['Avg. Session Length', 'Time on App', 'Time on Website', 'Length of Membership']]
y = df['Yearly Amount Spent']
X_train, X_test, y_train, y_test = train_test_split(X , y, test_size=0.3, random_state = 42)
```

```python
# Training the model
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
lm.fit(X_train, y_train) # They are fitted to Linear Regression
```

```python
# Showing Coefficients of each feature (independent variable)
cdf = pd.DataFrame(lm.coef_, X.columns, columns = ['Coef'])
print(cdf)
```

```python
# Predictions
preds = lm.predict(X_test)
preds
```

```python
# Graphical Evaluation of Predictions
sns.scatterplot(x = preds, y = y_test, alpha = 0.5)
plt.xlabel('Predictions (predicted values)')
plt.ylabel('Yearly Amount Spent (actual values)')
plt.title('Evaluation of our LM model')
```

```python
# Analytical evaluation of errors
from sklearn.metrics import mean_squared_error, mean_absolute_error
import math
```

```python
print('Mean Absolute Error: ', mean_absolute_error(y_test, preds))
print('Mean Squared Error: ', mean_squared_error(y_test, preds))
print('RMS Error: ', math.sqrt(mean_squared_error(y_test, preds)))
```

```python
# Residual analysis
residuals = y_test - preds
residuals # these should be random, otherwise the model can be biased
```

```python
sns.displot(residuals, color = 'red', bins = 15, kde = True)
```

```python
# Probability plot to show that residuals are normally distributed
import pylab
import scipy.stats as stats
stats.probplot(residuals, dist = 'norm', plot = pylab)
pylab.show()
```
